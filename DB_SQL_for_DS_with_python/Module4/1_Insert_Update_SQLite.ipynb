# Create & Access SQLite database using Python
## Task 1: Create database using SQLite

```python
import sqlite3
# connection to the database
conn = sqlite3.connect('INSTRUCTOR.db')
# create a cursor object to interact with the database
cursor_obj = conn.cursor()
```

## Task 2: Create a table in the database

In this step we will create a table in the database with following details:

| Column name | Data type | Nullable |
|-------------|-----------|----------|
|      ID     |  Integer  |     N    |
|    FNAME    |  Varchar  |     Y    |
|    LNAME    |  Varchar  |     Y    |
|     CITY    |  Varchar  |     Y    |
|    CCODE    | Character |     Y    |


```python
# Drop the table if already exists.
cursor_obj.execute("DROP TABLE IF EXISTS INSTRUCTOR")
```

**Activate the database**

Terminal: `sqlite3 INSTRUCTOR.db`

```python
# Creating table
table = """ create table IF NOT EXISTS INSTRUCTOR(ID INTEGER PRIMARY KEY NOT NULL, 
                                                  FNAME VARCHAR(20), 
                                                  LNAME VARCHAR(20), 
                                                  CITY VARCHAR(20), 
                                                  CCODE CHAR(2));"""
cursor_obj.execute(table) 
print("Table is Ready")
```

## Task 3: Insert data into the table

```python
# inserting single line
cursor_obj.execute('''insert into INSTRUCTOR values (1, 'Rav', 'Ahuja', 'TORONTO', 'CA')''')
```

```python
# inserting multiple lines
cursor_obj.execute('''insert into INSTRUCTOR values (2, 'Raul', 'Chong', 'Markham', 'CA'), (3, 'Hima', 'Vasudevan',  'Chicago', 'US')''')
```

## Task4: Query to explore data

```python
# stores sql as a string
sql = '''select * 
         from INSTRUCTOR'''
# sends the sql to the database using the cursor (PS.but no data returned yet)
cursor_obj.execute(sql)
print("Selected all data successfully")
# retrieves all rows produced by the query
output_all = cursor_obj.fetchall()
# iterates through each row returned from the database
for row_all in output_all:
    print(row_all)
```

```python
# 1- select specific rows
sql = '''select * 
         from INSTRUCTOR'''
cursor_obj.execute(sql)
print("Selected specific rows successfully")

output_specific = cursor_obj.fetchmany(2)
for row_specific in output_specific:
    print(row_specific)
```

```python
# 2- select specific columns
sql = '''select fname
         from INSTRUCTOR'''
cursor_obj.execute(sql)
print("Selected only fname column successfully")

output_fname = cursor_obj.fetchall()
for row_fname in output_fname:
    print(row_fname)
```

```python
# 3- write and execute an update with sql that changes the Rav's CITY to MOOSETOWN
sql = '''update instructor
         set city = 'MOOSETOWN'
         where fname = 'Rav';'''
cursor_obj.execute(sql)
print("Updated Rav's city successfully")

sql_verify = '''select *
                from INSTRUCTOR'''
cursor_obj.execute(sql_verify)
output_current = cursor_obj.fetchall()
for row_current in output_current:
    print(row_current)
```
